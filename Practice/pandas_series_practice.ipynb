```python
import pandas as pd
```

```python
# Pandas Series
data = [10, 20, 30, 40, 50]
series = pd.Series(data)
print("Pandas Series:")
print(series)
```

```python
# Pandas Indexing and Slicing
print("\nIndexing and Slicing:")
print("Element at index 2:", series[2])
print("Elements from index 1 to 3:")
print(series[1:4])
```

```python
# Pandas boolean indexing
print("\nBoolean Indexing:")
print("Elements greater than 25:")
print(series[series > 25])
```

```python
# Pandas Logical indexing
print("\nLogical Indexing:")
print("Elements greater than 20 and less than 50:")
print(series[(series > 20) & (series < 50)])
```

```python
# Pandas isnull() and notnull()
print("\nNull Value Handling:")
data_with_nulls = [10, None, 30, None, 50]
series_with_nulls = pd.Series(data_with_nulls)
print("Series with null values:")
print(series_with_nulls)
print("Null values in the series:")
print(series_with_nulls.isnull())
print("Non-null values in the series:")
print(series_with_nulls.notnull())
```

```python
# Pandas fillna() and dropna()
print("\nHandling Null Values:")
print("Series with null values:")
print(series_with_nulls)
print("Series after filling null values with 0:")
print(series_with_nulls.fillna(0))
print("Series after dropping null values:")
print(series_with_nulls.dropna())
```

```python
# Pandas Operations on Series
print("\nOperations on Series:")
print("Original Series:")
print(series)
print("Series after adding 5:")
print(series + 5)
print("Series after multiplying by 2:")
print(series * 2)
print("Series after applying a lambda function (squared):")
print(series.apply(lambda x: x**2))
```

```python
# Pandas Series with custom index
print("\nSeries with Custom Index:")
custom_index_series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print("Series with custom index:")
print(custom_index_series)
print("Element at index 'c':", custom_index_series['c'])
print("Elements from index 'b' to 'd':")
print(custom_index_series['b':'d'])
```

```python
# Pandas Series with datetime index
print("\nSeries with Datetime Index:")
dates = pd.date_range('20230101', periods=5)
datetime_index_series = pd.Series(data, index=dates)
print("Series with datetime index:")
print(datetime_index_series)
print("Element at index '2023-01-03':", datetime_index_series['2023-01-03'])
print("Elements from index '2023-01-02' to '2023-01-04':")
print(datetime_index_series['2023-01-02':'2023-01-04'])
```

```python
# Pandas Aggregation functions
print("\nAggregation Functions:")
print("Original Series:")
print(series)
print("Sum of the series:", series.sum())
print("Minimum value in the series:", series.min())
print("Maximum value in the series:", series.max())
print("Count of non-null values in the series:", series.count())
print("Unique values in the series:", series.unique())
print("Mean of the series:", series.mean())
print("Median of the series:", series.median())
print("Mode of the series:", series.mode())
print("Standard deviation of the series:", series.std())
print("Variance of the series:", series.var())
print("Value counts in the series:")
print(series.value_counts())
print("Describe the series:")
print(series.describe())
```

```python

print("Cumulative sum of the series:")
print(series.cumsum())
print("Cumulative product of the series:")
print(series.cumprod())
print("Cumulative maximum of the series:")
print(series.cummax())
print("Cumulative minimum of the series:")
print(series.cummin())
print("Quantiles of the series:")
print(series.quantile([0.25, 0.5, 0.75]))
print("Correlation with another series:")
other_series = pd.Series([5, 15, 25, 35, 45])
print(series.corr(other_series))
print("Covariance with another series:")
print(series.cov(other_series))
print("Skewness of the series:", series.skew())
print("Kurtosis of the series:", series.kurt())
```

```python
# Pandas Method Chaining
print("\nMethod Chaining:")
result = (series
          .apply(lambda x: x * 2)
          .apply(lambda x: x + 5)
          .apply(lambda x: x**2))
print("Result of method chaining:")
print(result)
```

```python
# Monthly temperature data for a year
print("\nMonthly Temperature Data:")
temperature_data = [30, 32, 28, 25, 27, 29, 31, 33, 26, 24, 22, 30]
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
temperature_series = pd.Series(temperature_data, index=months)
print(temperature_series)
mean_temperature = temperature_series.mean()
print("\nMean Temperature:", mean_temperature)
# Monthly temperature greater thabn mean
print("\nMonths with temperature greater than mean:")
print(temperature_series[temperature_series > mean_temperature])
# Categorize temperatures
print("\nCategorized Temperatures:")
def categorize_temperature(temp):
    if temp < 25:
        return 'Cold'
    elif 25 <= temp < 30:
        return 'Warm'
    else:
        return 'Hot'
temperature_categories = temperature_series.apply(categorize_temperature)
print(temperature_categories)
```

```python
# Pandas String Methods
print("\nPandas String Methods:")
string_data = ['Hello', 'World', 'Pandas', 'Series', 'Practice']
string_series = pd.Series(string_data)
print("Original String Series:")
print(string_series)
print("\nString Series in lowercase:")
print(string_series.str.lower())
print("\nString Series in uppercase:")
print(string_series.str.upper())
print("\nString Series with length of each string:")
print(string_series.str.len())
print("\nString Series with 'a' replaced by '@':")
print(string_series.str.replace('a', '@'))
print("\nString Series with 'Pandas' replaced by 'DataFrame':")
print(string_series.str.replace('Pandas', 'DataFrame'))
print("\nRemoving whitespace from String Series:")
print(string_series.str.strip())
print("\nChecking if String Series contains 'a':")
print(string_series.str.contains('a'))
print("\nSplitting String Series by 'o':")
print(string_series.str.split('o'))
print("\nExtracting first 3 characters from String Series:")
print(string_series.str[:3])
print("\nTitle case String Series:")
print(string_series.str.title())
print("\nString Series with regex replacement (replace vowels with '*'):")
print(string_series.str.replace(r'[aeiouAEIOU]', '*', regex=True))
```
