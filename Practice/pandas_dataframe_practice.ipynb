```python
import pandas as pd
```

```python
# Pandas DataFrame Practice
# Create a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df = pd.DataFrame(data)
print(df)
# Accessing columns
print(df['Name'])
# Accessing rows using loc
print(df.loc[0])  # First row
print(df.loc[1:3])  # Rows from index 1 to 3
# Accessing rows using iloc
print(df.iloc[0])  # First row
print(df.iloc[1:3])  # Rows from index 1 to 3
# Adding a new column
df['Salary'] = [50000, 60000, 70000, 80000]
print(df)
# Filtering data
filtered_df = df[df['Age'] > 30]
print(filtered_df)
# Grouping data
grouped_df = df.groupby('City')
print(grouped_df)
# Sorting data
sorted_df = df.sort_values(by='Age')
print(sorted_df)
```

```python
# Handling missing data
df_with_nan = df.copy()
df_with_nan.loc[2, 'Age'] = None
print(df_with_nan)
print(df_with_nan.isnull())
print(df_with_nan.fillna(0))
print(df_with_nan.dropna())
```

```python
# Querying in DataFrame
query_result = df.query('Age > 30')
print(query_result)
print(df.query('City == "New York"'))
print(df.query('Salary >= 60000'))
print(df.query('Age > 35 and Salary >= 70000'))
```

```python
# DataFrame operations
# Merging DataFrames
data2 = {
    'Name': ['Eve', 'Frank'],
    'Age': [28, 33],
    'City': ['San Francisco', 'Seattle']
}
df2 = pd.DataFrame(data2)
merged_df = pd.concat([df, df2], ignore_index=True)
print(merged_df)
# Pivoting DataFrame
pivot_df = df.pivot_table(values='Salary', index='City', aggfunc='mean')
print(pivot_df)
# Reshaping DataFrame
melted_df = df.melt(id_vars=['Name'], value_vars=['Age', 'Salary'])
print(melted_df)
pivoted_back_df = melted_df.pivot(index='Name', columns='variable', values='value')
print(pivoted_back_df)

```

```python
# Wrangling DataFrame
# Handling duplicates
df_with_duplicates = pd.concat([df, df.iloc[[0]]], ignore_index=True)
print(df_with_duplicates)
print(df_with_duplicates.duplicated())
print(df_with_duplicates.drop_duplicates())
# Handling outliers
df_with_outliers = df.copy()
df_with_outliers.loc[0, 'Salary'] = 1000000  # Adding an outlier
print(df_with_outliers)
print(df_with_outliers.describe())
# Handling categorical data
df_with_categorical = df.copy()
df_with_categorical['City'] = df_with_categorical['City'].astype('category')
print(df_with_categorical)
print(df_with_categorical['City'].cat.categories)
print(df_with_categorical['City'].cat.codes)
```
