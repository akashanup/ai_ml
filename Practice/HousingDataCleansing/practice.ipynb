```python
"""
As you can see, there is a column in the Attribute Dataset named 'Size’. This column contains the values in an abbreviated format. Write a code in Python to convert the following:


M into “Medium”

L into “Large”

XL into “Extra large”

Free into “Free”

S, s and small into “Small”

Now, once you are done with changes in the dataset, what is the value of the lowest percentage, the highest percentage and the percentage of Small size categories in the column named “Size”, respectively?
"""
# Replace values in the 'Size' column
df['Size'] = df['Size'].replace({
    'M': 'Medium',
    'L': 'Large',
    'XL': 'Extra large',
    'Free': 'Free',
    'S': 'Small',
    's': 'Small',
    'small': 'Small'
})
# Calculate the percentage of each category in the 'Size' column
size_percentages = df['Size'].value_counts(normalize=True) * 100
# Get the lowest percentage, highest percentage, and percentage of Small size categories
lowest_percentage = size_percentages.min()
highest_percentage = size_percentages.max()
small_percentage = size_percentages.get('Small', 0)
print(f"Lowest percentage: {lowest_percentage:.2f}%")
print(f"Highest percentage: {highest_percentage:.2f}%")
print(f"Percentage of Small size categories: {small_percentage:.2f}%")
```

```python
"""
You are given another dataset named “Dress Sales”. Now if you observe the datatypes of the columns using ‘inp1.info()’ command, you can identify that there are certain columns defined as object data type though they primarily consist of numeric data.


Now, if you try and convert these object data type columns into numeric data type(float), you will come across an error message. Choose the correct option with the reason for the error and also how to remove this error.
"""
# Load the Dress Sales dataset
inp1 = pd.read_csv('Dress+Sales (1).csv')
# Check the data types of the columns
print(inp1.info())
# Convert all object columns to numeric safely
object_cols = inp1.select_dtypes(include='object').columns
print("Object columns before conversion:", list(object_cols))

for col in object_cols:
    #inp1[col] = pd.to_numeric(inp1[col])  # non-numeric values -> NaN
    inp1[col] = pd.to_numeric(inp1[col], errors='coerce')  # non-numeric values -> NaN

# Check the data types again after conversion
print(inp1.info())
```

```python
"""

You need to categorise the dates into seasons in “Dress Sales” data to simplify the analysis according to the following criteria:


June, July and August: Summer.

September, October and November: Autumn.

December, January and February: Winter.

March, April and May: Spring.

Which of the seasons has the lowest sales among all the seasons, and what is its value?
"""
# Convert the 'Date' column to datetime format
inp1['Date'] = pd.to_datetime(inp1['Date'], errors='coerce')
# Define a function to categorize dates into seasons
def categorize_season(date):
    if pd.isnull(date):
        return 'Unknown'
    month = date.month
    if month in [6, 7, 8]:
        return 'Summer'
    elif month in [9, 10, 11]:
        return 'Autumn'
    elif month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
# Apply the function to create a new 'Season' column
inp1['Season'] = inp1['Date'].apply(categorize_season)
# Calculate total sales for each season
season_sales = inp1.groupby('Season')['Sales'].sum()
# Find the season with the lowest sales
lowest_sales_season = season_sales.idxmin()
lowest_sales_value = season_sales.min()
print(f"Season with the lowest sales: {lowest_sales_season}")
print(f"Lowest sales value: {lowest_sales_value}")

```
