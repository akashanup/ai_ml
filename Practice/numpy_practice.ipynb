```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4, 5])
print("1D array:")
print(a)
print("Shape of a:", a.shape)
print("Data type of a:", a.dtype)
print("Number of dimensions of a:", a.ndim)
print("Size of a:", a.size)
```

```python
# 2D array
b = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array:")
print(b)
print("Shape of b:", b.shape)
print("Data type of b:", b.dtype)
print("Number of dimensions of b:", b.ndim)
print("Size of b:", b.size)
```

```python
# 3D array  
c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print("3D array:")
print("Shape of c:", c.shape)
print("Data type of c:", c.dtype)
print("Number of dimensions of c:", c.ndim)
print("Size of c:", c.size)
print(c)
```

```python
# Array creation
d = np.zeros((2, 3))
print("Array of zeros:")
print(d)
e = np.ones((3, 4))
print("Array of ones:")
print(e)
f = np.full((2, 2), 7)  
print("Array filled with 7:")
print(f)
g = np.eye(4)
print("Identity matrix:")
print(g)
h = np.random.rand(2, 3)
print("Random array:")
print(h)
h1 = np.random.uniform(low=0, high=10, size=(2, 3))
print("Random array with uniform distribution:")
print(h1)
```

```python
# Array operations
i = np.array([1, 2, 3])
j = np.array([4, 5, 6])
print("Element-wise addition:")
print(i + j)
print("Element-wise multiplication:")
print(i * j)
print("Dot product:")
print(np.dot(i, j))
print("Transpose of a 2D array:")
k = np.array([[1, 2], [3, 4]])
print(k.T)
```

```python
# Array slicing and indexing
l = np.array([10, 20, 30, 40, 50])
print("Slicing from index 1 to 3:")
print(l[1:4])
print("Slicing with step of 2:")
print(l[::2])
print("Negative indexing:")
print(l[-2:])
print("Boolean indexing:")
m = np.array([1, 2, 3, 4, 5])
print("Elements greater than 3:")
print(m[m > 3])
```

```python
# Reshaping and flattening
n = np.array([[1, 2, 3], [4, 5, 6]])
print("Original array:")
print(n)
print("Reshaped array (3, 2):")
print(n.reshape(3, 2))
print("Flattened array:")
print(n.flatten())
```

```python
# Statistical operations
o = np.array([1, 2, 3, 4, 5])
print("Mean:", np.mean(o))
print("Median:", np.median(o))  
print("Standard deviation:", np.std(o))
print("Variance:", np.var(o))

```

```python
# Broadcasting: Adding a 1D array to a 2D array, where the 1D array is broadcasted across the rows of the 2D array.
p = np.array([1, 2, 3])
q = np.array([[1], [2], [3]])
print("Original arrays:")
print("p:", p)
print("q:\n", q)
print("Broadcasted addition:")
print(p + q)
```

```python
# Image as a 2D array (grayscale) and 3D array (color)
r = np.array([[0, 255], [255, 0]])
print("Grayscale image as a 2D array:")
print(r)

# Color image as a 3D array (RGB)
color_image = np.array([[[255, 0, 0], [0, 255, 0]], [[0, 0, 255], [255, 255, 0]]])
print("Color image as a 3D array:")
print(color_image)
```

```python
# Read image using PIL and convert to NumPy array
from PIL import Image
try:
    img = Image.open('images.jpeg')  # Replace with your image path
    img_array = np.array(img)
    print("Image as a NumPy array:")
    print(img_array)
    print("Shape of the image array:", img_array.shape)
    print("Data type of the image array:", img_array.dtype)
    print("Number of dimensions of the image array:", img_array.ndim)
    print("Size of the image array:", img_array.size)

    # Invert the colors of the image
    inverted_img_array = 255 - img_array
    inverted_img = Image.fromarray(inverted_img_array)
    inverted_img.save('inverted_image.jpeg')  # Save the inverted image

    # Convert the image to grayscale
    grayscale_img = img.convert('L')
    grayscale_img.save('grayscale_image.jpeg')  # Save the grayscale image
except FileNotFoundError:
    print("Image file not found. Please check the path and try again.")

```

```python
# Kernel or Filter application
# Applying a simple edge detection kernel to a given image using PIL and SciPy (if available)
try:
    from scipy import ndimage
    # Define a simple edge detection kernel (Sobel operator)
    kernel = np.array([[-1, 0, 1],
                       [-2, 0, 2],
                       [-1, 0, 1]], dtype=np.float32)

    # Reuse existing grayscale_array if available; otherwise create it
    if 'grayscale_array' not in globals():
        img = Image.open('images.jpeg')  # Replace with your image path
        grayscale_img = img.convert('L')
        grayscale_array = np.array(grayscale_img)

    # Apply the 2D kernel to the 2D grayscale image
    edges = ndimage.convolve(grayscale_array.astype(np.float32), kernel, mode='reflect')
    edges = np.clip(np.abs(edges), 0, 255).astype(np.uint8)
    
    # Save the edge-detected image
    edges_img = Image.fromarray(edges)
    edges_img.save('edges_image.jpeg')
except ImportError:
    print("Scipy is not installed. Please install it to use the edge detection feature.")
except FileNotFoundError:
    print("Image file not found. Please check the path and try again.")
```
